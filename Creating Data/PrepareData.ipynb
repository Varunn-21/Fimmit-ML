```python
import pandas as pd 
import numpy as np
```

```python
import requests
import json
data = []
url = "https://api.github.com/search/repositories"
for page in range(1,6):
    params = {"q": "language:javascript",
              "per_page": 100,
              "page": page,}
    
    response = requests.get(url, params=params)
    d = response.json()
    data.extend(d['items'])
```

```python
# data = data["items"]
for i in data[0]:
    print(i['name'])
```

```python
repos = []
for item in data:
    repos.append([
        item['description'],
        item['name'],
        listToStr(item['language']),
        age(item["created_at"]),
        listToStr(item['topics'])
    ])

df = pd.DataFrame(repos,columns=[["Description" , "Name" , "Language","Account_age" ,"Label"]])
```

```python
def listToStr(L):
    if type(L) is list:
        str = ''
        for i in L:
            str = str + " " + i
        return str
    else:
        return L
```

```python
from datetime import datetime, timezone
def age(date):
    created = datetime.fromisoformat(date.replace("Z", "+00:00"))
    age_years = (datetime.now(timezone.utc) - created).days / 365
    return round(age_years)
```
